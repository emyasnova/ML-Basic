$$

-1 & y & 4
\end{pmatrix}

2 & -6 & z
\end{pmatrix}
= \begin{pmatrix}

$ \\

### Задание 8.


В задании 5 подберите прямую из множества **всех** прямых на плоскости, используя матричное решение метода наименьших квадратов: $w_{min} = (X^TX)^{-1}X^Ty$.

```python
import numpy as np
import matplotlib.pyplot as plt


X = np.array([
    [1, -3],
    [1, 2],
    [1, 5]
])
y = np.array([2, 2, -3])
Xt = X.T

b, a = np.linalg.inv(Xt@X)@Xt@y


x = np.linspace(-10, 10, 100)
y = a*x+b
plt.plot(x, y, label="Минимальная сумма отклонений")

plt.scatter(-3, 2)
plt.scatter(2, 2)
plt.scatter(5, -3)
plt.grid(True)
plt.legend()
plt.show()
```
